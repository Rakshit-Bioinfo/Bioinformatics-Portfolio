```python
patient_reads = ["ATGCGTAAC", "ATC", "TTAACCGGC"]
CODON_TABLE = {
    "AUG": "Methione",
    "UUA": "Leucine"
}

for dna in patient_reads:
    if len(dna) < 6:
        print(dna, "--> FAILED: Read is too short.")
    elif len(dna) > 9:
        print(dna, "--> FAILED: Read is too long.")
    else:
        rna = dna.replace("T", "U")
        # Take the first codon (first 3 bases)
        first_codon = rna[:3]
        amino_acid = CODON_TABLE.get(first_codon, "Unknown")
        print(dna, f"--> PASSED! First amino acid is: {amino_acid}")
```

```python
patient_samples = ["AATGCGT", "TGC", "GCTAGCA"]
disease_db = {
    "AAT": "virus A",

    "GCT": "virus B"
}
for sample in patient_samples:
  if len(sample) < 5:
    print(sample, "--> FAILED: Sample is too short.")
  elif len(sample) > 10:
    print(sample, "--> FAILED: Sample is too long.")
  else:
    first_three = sample[0:3]
    diagnosis = disease_db.get(first_three, "HEALTHY")
    print(sample, "--> Diagnosis:", diagnosis)
```
